# A-MESH example

A-MESH minimizes every objective returned by the objective function. The current Python API is imported from `amesh`.

```python
import numpy as np

from amesh import AMESH, AMESHParameters
```

The example below minimizes two conflicting objectives over five bounded decision variables.

```python
def objectives(x: np.ndarray) -> np.ndarray:
    return np.array([
        np.sum(x**2),
        np.sum((x - 1.0) ** 2),
    ])


decision_dim = 5
params = AMESHParameters(
    objective_dim=2,
    decision_dim=decision_dim,
    decision_lower_bounds=np.full(decision_dim, -5.0),
    decision_upper_bounds=np.full(decision_dim, 5.0),
    population_size=40,
    max_fit_eval=4_000,
    random_state=42,
)

amesh = AMESH(params, objectives)
amesh.run()
positions, objective_values = amesh.get_results()
```

The returned arrays contain the retained non-dominated decision vectors and their corresponding objective values.
